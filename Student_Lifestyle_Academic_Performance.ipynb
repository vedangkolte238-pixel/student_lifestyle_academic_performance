## 9. Multivariate Analysis

Multivariate analysis examines three or more variables together. A correlation heatmap is used to study the relationships among numerical variables and identify variables that have stronger positive or negative associations with academic performance.

```python
# Visualization 6: Correlation heatmap
# Variables is more than 2 variables called multivariate correlation.
# For multivariate correlation, we can use a heatmap to visualize the correlation matrix of numerical variables in the dataset.
numeric_df = df.select_dtypes(include=np.number)

# taking the size of the figure to 10 inches wide and 7 inches tall.
plt.figure(figsize=(10, 7))

# Creating a heatmap to display the correlation matrix of numerical variables in the dataset.
# The heatmap uses a color gradient to represent the strength of the correlation between variables.
# Red = Strong positive correlation, Blue = Strong negative correlation, White = No correlation.
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)

# Adding title to the plot.
plt.title("Correlation Heatmap of Numerical Variables")

# Displaying the plot.
plt.show()
```

### Interpretation

## 10. Outlier Analysis

Outlier analysis is used to identify unusually high or low observations. The Interquartile Range (IQR) method is used to identify potential outliers, and a boxplot is used for visual inspection.

```python
# Identify outliers using the IQR method.
outlier_summary = {}

# taking loop for each column to identify outliers.
for column in ["Study_Hours", "Sleep_Hours", "Screen_Time", "Attendance", "Social_Time", "Previous_Score", "Academic_Score"]:
    Q1 = df[column].quantile(0.25) # Q1 = 0.25% of the data
    Q3 = df[column].quantile(0.75) # Q3 = 0.75% of the data
    IQR = Q3 - Q1  # Subtracting Q1 from Q3 to get 50% of the data.
    lower = Q1 - 1.5 * IQR # For lower bound, we subtract 1.5 times the IQR from Q1.
    upper = Q3 + 1.5 * IQR # For upper bound, we add 1.5 times the IQR to Q3.

    # Here we mention model that which columns comes in upper and loweer bound.
    # Also we sum the count of outliers in each column and store it in the outlier_summary dictionary.
    count = ((df[column] < lower) | (df[column] > upper)).sum()
    outlier_summary[column] = count

# Printing the summary of outliers for each column to see the results.
print(outlier_summary)
```

```python
# Visualization 7: Academic score outliers
# taking the size of the figure to 8 inches wide and 4 inches tall.
plt.figure(figsize=(8, 4))

# Creating a boxplot to visualize the distribution of academic scores and identify potential outliers.
sns.boxplot(x=df["Academic_Score"])

# Adding title to the plot.
plt.title("Boxplot of Academic Scores for Outlier Detection")

# Adding x label(name) for the plot.
plt.xlabel("Academic Score")

# Displaying the plot.
plt.show()
```

### Outlier Interpretation

The boxplot and IQR results help identify potential unusual academic scores. Potential outliers should not automatically be removed because they may represent genuine student observations. They should be investigated before deciding whether to retain or handle them.

## 11. Key Insights

The following code calculates important findings from the dataset instead of relying only on visual observation.

```python
# Calculate key statistics for insights.
# Adding correlation to each column contain variables.
study_corr = df["Study_Hours"].corr(df["Academic_Score"])
screen_corr = df["Screen_Time"].corr(df["Academic_Score"])
sleep_corr = df["Sleep_Hours"].corr(df["Academic_Score"])
attendance_corr = df["Attendance"].corr(df["Academic_Score"])

# Grouping the data by lifestyle category and calculating the average academic score for each lifestyle category.
lifestyle_scores = df.groupby("Lifestyle")["Academic_Score"].mean().sort_values(ascending=False)

# Displaying the calculated statistics and insights.
# Using f-string formatting to display the correlation values rounded to 3 decimal places.
print(f"1. Study Hours vs Academic Score correlation: {study_corr:.3f}")
print(f"2. Screen Time vs Academic Score correlation: {screen_corr:.3f}") # takes 3 decimal numbers after decimal point.
print(f"3. Sleep Hours vs Academic Score correlation: {sleep_corr:.3f}")
print(f"4. Attendance vs Academic Score correlation: {attendance_corr:.3f}")
print("\n5. Average Academic Score by Lifestyle:")
print(lifestyle_scores.round(2))
print(f"\n6. Average Academic Score: {df['Academic_Score'].mean():.2f}")
print(f"7. Average Study Hours: {df['Study_Hours'].mean():.2f} hours/day")
```

### Key Findings from the Dataset

- Study hours show a positive relationship with academic score in this dataset.
- Screen time shows a negative relationship with academic score.
- Sleep hours and attendance show positive relationships with academic score, although their relationships are weaker than study hours.
- Lifestyle categories have different average academic scores, which can be compared using the group analysis.
- The dataset contains some potential outliers identified using the IQR method, but these should be investigated rather than automatically removed.
